```python
from sympy import symbols, Symbol, I
from sympy.matrices import Matrix
import sympy as sp
```

```python
# p0 = a
# p1 = b
# p2 = c
# m  = m
a = symbols("p_0", real=True, positive=True)
b, c = symbols("p_1, p_2", real=True)
m = Symbol("m", real=True, positive=True)
x, y = symbols("x, y", real=True)
zero = Matrix([0, 0])

u_matrix = Matrix([[a - m, -b - I * c],
                   [b - I * c, -a - m]])
u_matrix = u_matrix.subs(a, sp.sqrt(b ** 2 + c ** 2 + m ** 2))
u_matrix
```

```python
u_sol = sp.linsolve((u_matrix, zero),
                    [x, y]
                    )
u_sol
```

```python
v_matrix = Matrix([[a + m, -b - I * c],
                   [b - I * c, -a + m]])
v_matrix = v_matrix.subs(a, sp.sqrt(b ** 2 + c ** 2 + m ** 2))
v_matrix
```

```python
v_sol = sp.linsolve((v_matrix, zero),
                    [x, y]
                    )
v_sol
```

```python
u = Matrix(*u_sol)
v = Matrix(*v_sol)
u_dagger = u.transpose().conjugate()
v_dagger = v.transpose().conjugate()
u_bar = u_dagger * Matrix([[1, 0],
                           [0, -1]])
v_bar = v_dagger * Matrix([[1, 0],
                           [0, -1]])
```

# Norm u

```python
test_u_norm = u_bar * u
test_u_norm.simplify()
test_u_norm[0, 0]
```

```python
y_2_u = 2 * m / (test_u_norm[0, 0] / y ** 2)
y_2_u
```

```python
test_u_norm.subs(y ** 2, y_2_u)[0, 0]
```

This choice of norm results in the usual convention of $\bar{u} u = 2m$

```python
test = (u_dagger * u).subs(y ** 2, y_2_u)
test.simplify()
test[0, 0]
```

This choice of norm results in the usual convention of $u^\dagger u = 2E_p$

```python
u = u.subs(y, sp.sqrt(y_2_u))
u.simplify()
u_dagger = u_dagger.subs(y, sp.sqrt(y_2_u))
u_dagger.simplify()
u_bar = u_bar.subs(y, sp.sqrt(y_2_u))
u_bar.simplify()
```

# Norm v

```python
test_v_norm = v_bar * v
test_v_norm.simplify()
test_v_norm[0, 0]
```

```python
y_2_v = -2 * m / (test_v_norm[0, 0] / y ** 2)
y_2_v
```

```python
test_v_norm.subs(y ** 2, y_2_v)[0, 0]
```

This choice of norm results in the usual convention of $\bar{v} v = -2m$

```python
test = (v_dagger * v).subs(y ** 2, y_2_v)
test.simplify()
test[0, 0]
```

This choice of norm results in the usual convention of $v^\dagger v = 2E_p$

```python
v = v.subs(y, sp.sqrt(y_2_v))
v.simplify()
v_dagger = v_dagger.subs(y, sp.sqrt(y_2_v))
v_dagger.simplify()
v_bar = v_bar.subs(y, sp.sqrt(y_2_v))
v_bar.simplify()
```

```python
import numpy as np
from qiskit_ibm_runtime.options.utils import UnsetType
from scipy import optimize
```

```python
optimize.show_options("minimize", "cobyla")
```

```python
def cost_func(x, a, b, c):
    # print(x)
    return a * (x[0]) ** 2 + b * (x[1]) ** 2 + c
optimizer_parameters = {
    "method": 'cobyla',
    "bounds": None,
    "constraints": (),
    "tol": None,
    "callback": None,
    "options": {"maxiter": 100,
                "disp": 2},
}
```

```python
# noinspection PyTypeChecker
optimize.minimize(fun=cost_func,
                  x0=np.array([2.3, 3.14]),
                  args=(1, 1, 1),
                  **optimizer_parameters
                  )
```

```python
class CostFunction:
    def __init__(self,  a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def __call__(self, x: np.ndarray) -> float:
        # print(x)
        return self.a * (x[0]) ** 2 + self.b * (x[1]) ** 2 + self.c
```

```python
optimize.minimize(fun=CostFunction(1, 1, 1),
                  x0=np.array([2.3, 3.14]),
                  **optimizer_parameters
                  )
```

```python
import inspect
from qiskit_ibm_runtime import EstimatorOptions
import qiskit_ibm_runtime.options as qiskit_ibm_options
b = EstimatorOptions()
```

```python
def check_for_option(value):
    temp = inspect.getmembers(qiskit_ibm_options, inspect.isclass)
    known_options = [x[1] for x in temp]
    result = False
    for option in known_options:
        result = result or isinstance(value, option)
    return result


def print_options(options, prefix=""):
    print(f"{prefix}{type(options).__name__}:")
    attrs = [a for a in dir(options) if not a.startswith('_') and not (callable(getattr(options, a)) or isinstance(getattr(options, a), UnsetType))]
    for attr in attrs:
        value = getattr(options, attr)
        if check_for_option(value):
            print_options(value, prefix=prefix+"\t")
        else:
            print(f"{prefix}\t{attr} = {value}")
print_options(b)
```

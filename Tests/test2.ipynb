```python
import numpy as np
from scipy import optimize
```

```python
optimize.show_options("minimize", "cobyla")
```

```python
def cost_func(x, a, b, c):
    # print(x)
    return a * (x[0]) ** 2 + b * (x[1]) ** 2 + c
```

```python
optimizer_parameters = {
    "method": 'cobyla',
    "bounds": None,
    "constraints": (),
    "options": {"maxiter": 100,
                "disp": 2},
}
# noinspection PyTypeChecker
optimize.minimize(fun=cost_func,
                  x0=np.array([2.3, 3.14]),
                  args=(1, 1, 1),
                  **optimizer_parameters
                  )
```
